### Only ICTV Headers

```python
import re

def process_header(header):
    parts = header.split('|')
    if len(parts) >= 2:
        last, second_last = parts[-1], parts[-2]
        parts = parts[:-2] + [last]

    return '/'.join(parts)

with open('meta2newfastanames_out\Astroviridae_complete_1B_noamb_mammalia_aln_gr.fna') as f, open('Astroviridae_complete_1B_noamb_mammalia_aln_gr_only_ictv_headers.fna', 'w') as out:
    for line in f:
        if line.startswith('>'):
            out.write('>' + process_header(line[1:].strip()) + '\n')
        else:
            out.write(line)
```

### GenBank Headers

```python
import re

def process_header(header):
    parts = header.split('|')
    if len(parts) >= 2:
        last, second_last = parts[-1], parts[-2]
        if last == 'NA' and second_last == 'NA':
            parts = parts[:-1]
        elif last == 'NA':
            parts = parts[:-2] + [second_last]
        elif second_last == 'NA':
            parts = parts[:-2] + [last]
        else:
            parts = parts[:-2] + [second_last]
    return '/'.join(parts)

with open('meta2newfastanames_out\Astroviridae_complete_1B_noamb_mammalia_aln.fna') as f, open('Astroviridae_complete_1B_noamb_mammalia_aln_genbank_headers.fna', 'w') as out:
    for line in f:
        if line.startswith('>'):
            out.write('>' + process_header(line[1:].strip()) + '\n')
        else:
            out.write(line)
```
