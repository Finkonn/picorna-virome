```python
ids = {l.strip().split("/")[0] for l in open("Mammalia_ids_all.txt") if l.strip()}

with open("Astroviridae_complete_1B_noamb.fna") as f, open("Astroviridae_complete_1B_noamb_mammalia.fna", "w") as out:
    keep = False
    for line in f:
        if line.startswith(">"):
            keep = line[1:].strip().split("/")[0] in ids
        if keep:
            out.write(line)
```
