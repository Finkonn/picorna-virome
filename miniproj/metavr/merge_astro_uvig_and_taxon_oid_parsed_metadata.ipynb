```python
import pandas as pd

df1 = pd.read_csv("metaVR_astroviridae_parsed_metadata.csv", sep=",", dtype=str)   
df2 = pd.read_csv("metaVR_astroviridae_uvigs.csv", sep=",", dtype=str)  

df1["Taxon ID"]  = df1["Taxon ID"].astype(str).str.strip()
df2["taxon_oid"] = df2["taxon_oid"].astype(str).str.strip()

merged = df2.merge(
    df1,
    left_on="taxon_oid",
    right_on="Taxon ID",
    how="left",         
    suffixes=("", "_ref")
)

out_cols = {
    "uvig":                     "UViG",
    "taxon_oid":                "Taxon_ID",
    "quality":                  "Quality",
    "completeness":             "Completeness",
    "cds_count":                "CDS_count",
    "length":                   "Length",
    "ictv_taxonomy":            "ICTV_taxonomy",
    "host_taxonomy":            "host_taxonomy",
    "Study Name":               "Study Name",
    "Ncbi Phylum":              "Ncbi Phylum",
    "Ncbi Class":               "Ncbi Class",
    "Ncbi Order":               "Ncbi Order",
    "Ncbi Family":              "Ncbi Family",
    "Ncbi Genus":               "Ncbi Genus",
    "Ncbi Species":             "Ncbi Species",
    "Host Name":                "Host Name",
    "Ncbi Genbank Id":          "Ncbi Genbank Id",
    "Geographic Location":      "Geographic Location",
    "Isolation Country":        "Isolation Country",
    "Sample Collection Date":   "Sample Collection Date",
    "Add Date":                 "Add Date",
    "Public Release Date":      "Public Release Date",
}

existing = {k: v for k, v in out_cols.items() if k in merged.columns}
result = merged[list(existing.keys())].rename(columns=existing)

#result.to_csv("metaVR_astroviridae_uvigs_metadata.csv", index=False, sep=",")  
result.to_excel("metaVR_astroviridae_uvigs_metadata.xlsx", index=False)

print(merged["Study Name"].isna().sum() if "Study Name" in merged else "n/a")
```
