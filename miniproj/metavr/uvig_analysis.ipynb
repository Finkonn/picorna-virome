```python
import pandas as pd
```

```python
df = pd.read_csv('uvig.tsv', sep='\t')
df.head()
```

```python
pattern = r'(?i)astrovir' 

astro_df = df[df['ictv_taxonomy'].str.contains(pattern, regex=True, na=False)].copy()
```

```python
astro_df.to_csv('astroviruses.csv', index=False)
#astro_df.to_excel('astroviruses.xlsx', index=False)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

astro_df['length'].hist(bins=50, ax=axes[0])
axes[0].set_title('Length distribution')
axes[0].set_xlabel('length')

astro_df['cds_count'].value_counts().sort_index().plot(kind='bar', ax=axes[1])
axes[1].set_title('CDS count distribution')
axes[1].set_xlabel('cds_count')

astro_df['quality'].value_counts().sort_index().plot(kind='bar', ax=axes[2])
axes[2].set_title('Quality distribution')
axes[2].set_xlabel('quality')

plt.tight_layout()
plt.savefig('astroviruses_stats.png', dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
astro_df['length'].hist(bins=50, edgecolor='black')
plt.title('Length distribution')
plt.xlabel('length')
plt.ylabel('count')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('length_dist.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
astro_df['cds_count'].value_counts().sort_index().plot(kind='bar', edgecolor='black')
plt.title('CDS count distribution')
plt.xlabel('cds_count')
plt.ylabel('count')
plt.grid(alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('cds_count_dist.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
astro_df['quality'].value_counts().sort_index().plot(kind='bar', edgecolor='black')
plt.title('Quality distribution')
plt.xlabel('quality')
plt.ylabel('count')
plt.grid(alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('quality_dist.png', dpi=150)
plt.show()
```
