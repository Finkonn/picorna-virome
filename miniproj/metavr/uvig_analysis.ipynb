```python
import pandas as pd
```

```python
df = pd.read_excel('metaVR_astroviridae_uvigs_metadata.xlsx')
df.head()
```

```python
df['Sample Collection Date'] = df['Sample Collection Date'].str.split('/').str[0]
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ensure the column is in datetime format
df['Sample Collection Date'] = pd.to_datetime(df['Sample Collection Date'])

# 2. Extract year, drop missing values, and convert to integer
years = df['Sample Collection Date'].dt.year.dropna().astype(int)

# 3. Count occurrences per year and sort sequentially
year_counts = years.value_counts().sort_index()

# 4. Plot as a bar chart
plt.figure(figsize=(10, 5))
year_counts.plot(kind='bar', edgecolor='black', width=0.8)

plt.title('Sample Collection Distribution by Year')
plt.xlabel('Year')
plt.ylabel('Count')

# Keep the labels completely horizontal
plt.xticks(rotation=45) 

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('year_dist.png', dpi=150)
plt.show()
```
